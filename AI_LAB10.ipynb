````python
import numpy as np

values = np.array([15,16,200,17,16,400,15,18])
print("Original values : " ,values)

mean_val = np.mean(values)
std_val = np.std(values)

print("Mean:", mean_val)
print("Std: ", std_val)

cleaned = np.where(values > 100 , mean_val , values)

print("Cleaned values : " ,cleaned)
````

## Lab Task 5 (15–20 minutes):
- Run the code and observe the original and cleaned values.
- Change threshold from 100 to 50 and observe what changes.
- Add one more wrong value at the end (e.g., 1) and modify the code so that:
  - Values > 100 OR values < 5 are replaced by mean.
- Write 3–4 lines in notebook: “If I don’t clean data before AI model, what problems can happen?”

````python
import numpy as np

values = np.array([15,16,200,17,16,400,15,18])
print("Original values : " ,values)

mean_val_1 = np.mean(values)
print("Mean:", mean_val_1)

# Change threshold from 100 to 50 and observe what changes.
cleaned2 = np.where(values > 50 , mean_val_1 , values)
print("Cleaned values  upto 50 : " ,cleaned2)

# Add one more wrong value at the end (e.g., 1)
values = np.array([15,16,200,17,16,400,15,18 , 1])

mean_val_2 = np.mean(values)
print("Mean:", mean_val_2)

#Values > 100 OR values < 5 are replaced by mean.
cleaned_3 = np.where((values > 100) | (values < 5) , mean_val_2  , values)
print("Cleaned Values > 100 OR values < 5 : " ,cleaned_3)
````
