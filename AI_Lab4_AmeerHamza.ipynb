```python
# Task 1
def get_grade(marks):
     # Use the conditions
     if marks >= 80:
         return "A"      # 80–100 → A
     elif marks >= 60:
         return "B"       # 60–79 → B
     elif marks >= 40:
         return "C"         # 40–59 → C
     else:
         return "Fail" # <40 → Fail

         

marks_list = [95, 72, 38, 60, 45]   #  data to demonstrate function usage

for m in marks_list:    # Loop through the list and print grade for each entry
 
    print(f"Marks: {m} → Grade: {get_grade(m)}")    # f-string shows both the marks and the computed grade
```

```python
# Task 2 

import string              # Provides string.punctuation (all common punctuation chars)

sentence = "Hello, Students! AI is amazing, isn't it?"     # Example input sentence; replace with input() to read from user if needed

sentence_lower = sentence.lower()          # 1) Lowercase conversion

cleaned = sentence_lower     # 2) Remove punctuation by replacing each punctuation char with empty string

for p in string.punctuation:  #string.punctuation is a function that take from string module  that remove punctustion from text
    cleaned = cleaned.replace(p , "")
    

words = cleaned.split()    # 3) Tokenization: split the cleaned sentence into words on whitespace


# Show results at each stage for clarity
print("Original:", sentence)
print("Lower :", sentence_lower)
print("Cleaned :", cleaned)
print("Words :", words)


```

```python
# Task 3 

import numpy as np # NumPy is the fundamental package for scientific computing 


arr=np.array([1,2,3,4,5,6,7])  #create a array

#perform basic operations on it
arr_sum=arr.sum()  #calculate a sum
arr_mean = arr.mean()  #calculate the mean of array
arr_max = arr.max()    #calculate the max value in array
arr_min = arr.min()   #calculate the min value in array

mask_above_mean  = arr > arr_mean  # Build a boolean mask: True where element is strictly greater than mean


above_mean_values = arr[mask_above_mean]   # Use the mask to filter elements


# Display all results
print("Array :", arr)
print("Sum :", arr_sum)
print("Mean :", arr_mean)
print("Max / Min :", arr_max, "/", arr_min)
print("Above Mean? :", mask_above_mean)         # shows True/False per element
print("Values > mean :", above_mean_values)

```

```python
# Task 5 

import numpy as np    

arr = np.arange(1, 21)               # Create integers from 1 to 20 (inclusive of 1, exclusive of 21)

even_mask = (arr % 2 == 0)   # Boolean masks: True where condition holds

odd_mask = ~even_mask # logical NOT of even_mask

# Filter arrays using masks
evens = arr[even_mask]
odds = arr[odd_mask]

# Count evens and odds using len(...) or .size
n_evens = len(evens)
n_odds = len(odds)


# Display all results
print("All numbers :", arr)
print("Even mask :", even_mask)
print("Odd mask :", odd_mask)
print("Evens :", evens, "→ count =", n_evens)
print("Odds :", odds, "→ count =", n_odds)
```

```python
# --- Wrap-Up Challenge Solution (Fully Commented) ---


import numpy as np    

np.random.seed(42)  # Set a random seed so results are repeatable (optional)

marks = np.random.randint(0, 101, size=10)   # Generate random marks for 10 students between 0 and 100 (inclusive of 0, exclusive of 100)

# Compute summary statistics
avg = marks.mean()
topper = marks.max()
lowest = marks.min()


 
above_avg_mask = marks > avg
above_avg_scores = marks[above_avg_mask]


# Display the full analysis
print("Marks array :", marks)
print("Class Average :", round(avg, 2))
print("Topper Score :", topper)
print("Lowest Score :", lowest)
print("Above-Average Mask :", above_avg_mask)
print("Above-Average Scores :", above_avg_scores)
```
