`sklearn.ensemble` is a module in the scikit-learn (sklearn) library that provides methods for **Ensemble Learning**.

“Combining many weak models to create one strong model.”

Example:  
Instead of taking 1 person’s opinion, you take the opinion of 100 people to make a better decision.

| Model Name | Type | What it Does |
|-----------|------|--------------|
| **RandomForestClassifier** | Classification | Builds many decision trees and makes a final decision using majority voting |
| **RandomForestRegressor** | Regression | Predicts values by averaging outputs of many trees |
| **GradientBoostingClassifier / Regressor** | Boosting | Learns by fixing errors step-by-step |
| **AdaBoost, BaggingClassifier** | Other ensemble methods | Combine multiple models using different strategies |

Because **Random Forest is itself an ensemble** —

```
🌲 + 🌲 + 🌲 + ... (100 trees) = 🌳🌳🌳  Random Forest
```

That’s why it lives inside the `ensemble` module — it uses a **group of models**, not just one.

#  7 Encoding Categorical Data

#### Why Encode?

Machine learning models work with numbers, not raw text. So we convert text columns (like city names) into numeric codes.

**7.1 Label Encoding**

`````python
from sklearn.preprocessing import LabelEncoder
`````

✅ What does it do?
It imports a tool (class) called `LabelEncoder` from the sklearn library, which converts text (string) data into numbers.

✅ Why is it used here?
Because machine learning models understand only numbers — not words like "Lahore" or "Karachi".
So we need to convert text columns like "city" into numbers.

✅ When do we use it?
When your data has categorical text features, such as:
city, country, color, brand, gender (male/female), etc.
Especially when there is an **order between categories** or you can assume one (e.g., low < medium < high).

⚠️ But if there is no order (e.g., Lahore, Islamabad, Karachi), Label Encoding may be weak — in that case, One-Hot Encoding is better (we will learn it later!).

`````python
le = LabelEncoder()
`````

✅ **What does it do?**  
It creates a new instance of the `LabelEncoder` class, named **le**.

✅ **Why is it used here?**  
So that we can use it later to encode our data.

✅ **When do we use it?**  
Every time you want to use a utility class (like `LabelEncoder`, `StandardScaler`), you first need to create its **object/instance**.


`````python
df_clean["city_encoded"] = le.fit_transform(df_clean["city"])

df_clean[["city" , "city_encoded"]].head()
`````

✅ **What does it do?**  
`le.fit_transform(...)` first maps all unique values in the **"city"** column (e.g., Lahore, Karachi, Islamabad...) to numbers like 0, 1, 2, ...  
Then it replaces each city in the dataframe with its corresponding number.  
A new column **"city_encoded"** stores these numeric values.  

**Example:**

| city       | city_encoded |
|------------|--------------|
| Karachi    | 0            |
| Lahore     | 1            |
| Islamabad  | 2            |

*Note:* The order may be alphabetical or based on occurrence — it depends on `LabelEncoder`.

✅ **Why is it used here?**  
So that the **"city"** column can be used by a machine learning model.

✅ **When do we use it?**  
- When you have **only one categorical feature**.  
- When there are **not too many unique values** (otherwise One-Hot Encoding is better).  
- When you are encoding **class labels (target)**, e.g., "Yes" → 1, "No" → 0 — Label Encoding is perfect for this case.


# 7.2 One-Hot Encoding with get_dummies()

`````python
df_dummies = pd.get_dummies(df_clean , columns=["city"])
df_dummies.head()
`````

✅ **What does it do?**  
It uses the **pandas** function `get_dummies()` which converts a categorical column (e.g., "city") into **binary (0 or 1) columns** — this is called **One-Hot Encoding**.  

**Example:**  
If the "city" column has values:  
"Karachi", "Lahore", "Islamabad"  

`get_dummies()` will create three new columns:  

- **city_Karachi** → 1 if the city is Karachi, else 0  
- **city_Lahore** → 1 if the city is Lahore, else 0  
- **city_Islamabad** → 1 if the city is Islamabad, else 0  

✅ **Why is it used here?**  
Machine learning models only understand numbers, and **Label Encoding** can mislead models for unordered categorical values (like cities) by implying a numeric order (e.g., 2 > 1 > 0).  
**One-Hot Encoding** solves this problem — each city becomes a separate feature, with no numeric hierarchy.  

✅ **When do we use it?**  
- When you have a **categorical feature** with **no natural order** (e.g., city, brand, color, country).  
- When the number of unique values is **reasonable** (usually 2–10; if 100+ values exist, it can make the dataset very wide — known as the "curse of dimensionality").  
- Especially useful for **linear models, SVMs, or neural networks**.  

✅ **Note:**  
Using `columns=["city"]` ensures only the "city" column is encoded — all other data remains unchanged.


# 🔹 Label Encoding vs One-Hot Encoding

## 1️⃣ Label Encoding

- Each category is assigned a **number**.  

**Example:**

| text       | number |
|------------|--------|
| Karachi    | 0      |
| Lahore     | 1      |
| Islamabad  | 2      |

**✔️ Advantage:**  
- Data size does **not increase** — only one column is needed.

**❌ Disadvantage:**  
- The model might assume an **order**: `2 > 1 > 0` → "Islamabad > Lahore > Karachi"  
- Which is **incorrect**, because cities have **no natural order**.

---

## 2️⃣ One-Hot Encoding

- Each category gets a **separate binary column** (0 or 1).

**Example:**

| city_Karachi | city_Lahore | city_Islamabad |
|--------------|-------------|----------------|
| 1            | 0           | 0              |
| 0            | 1           | 0              |
| 0            | 0           | 1              |

**✔️ Advantage:**  
- No incorrect relationships — each city is a separate feature.  
- The model gets **clear information**: "This individual is from Karachi, all others 0."

**❌ Disadvantage:**  
- If there are 100 cities, 100 new columns are created → **dataset becomes very wide** (**Curse of Dimensionality**).

---

## 🔸 Which is better?

❗ Answer: **It depends on the data!**

| Type                  | When to use?                                                                 | Why?                                                                 |
|-----------------------|-----------------------------------------------------------------------------|---------------------------------------------------------------------|
| ✅ Label Encoding       | When categories have a **natural order**                                     | e.g., Low < Medium < High, Poor < Fair < Good < Excellent          |
| ✅ One-Hot Encoding     | When categories have **no order**                                           | e.g., City, Country, Color, Brand, Gender                          |

---

## 🔹 Practical Examples

### Example 1: Ordered Categories → Label Encoding

**Feature:** `Education Level` = ["School", "Bachelor", "Master", "PhD"]  
- Order exists: `School < Bachelor < Master < PhD`  
- Use **Label Encoding**:  
  `School=0, Bachelor=1, Master=2, PhD=3`  
- 🤖 The model understands that `PhD > Bachelor` — **correct!**

### Example 2: Unordered Categories → One-Hot Encoding

**Feature:** `City` = ["Karachi", "Lahore", "Islamabad"]  
- No order exists  
- Use **One-Hot Encoding**  
- 🤖 If Label Encoding is used, the model assumes:  
  `"Islamabad (2) > Lahore (1)"` — **completely wrong!**

### Example 3: Target (y) Encoding → Label Encoding

**Feature:** `price_label` = ["Low", "High"]  
- Always use **Label Encoding** for the target, regardless of order.  
- Classification models need **numerical labels** (0, 1, 2…).

---

## 🔸 Additional Tips

| Situation                               | Recommendation                                                                 |
|----------------------------------------|-------------------------------------------------------------------------------|
| Unique values < 10                      | One-Hot Encoding is best                                                      |
| Unique values > 15–20                   | Consider Label Encoding or Target Encoding (avoid One-Hot)                   |
| In a data pipeline                      | Use `sklearn.preprocessing.OneHotEncoder` instead of `pd.get_dummies()` — supports `fit/transform` pattern |

---

## ✅ Summary (Decision Chart)

💡 Final Advice:  
**"When in doubt, use One-Hot Encoding — especially as a beginner."**  
- Safe  
- Easy to understand  
- Prevents sending the model in the wrong direction

  → Prevents false numerical relationships.

## 🔑 **Key Takeaways**
1. **Data workflow**: Load → Explore → Clean → Encode → Split → Train → Evaluate.
2. **Always split data** before training to avoid overfitting.
3. **Choose encoding wisely**:  
   - Ordered categories → Label Encoding  
   - Unordered categories → One-Hot Encoding
4. **Start simple**: Linear Regression for numbers, Random Forest for categories.
5. **Visualize early**: Histograms and bar charts reveal data issues quickly.
