# 🐍 Model in Python (OOP) — Extra Commented

LinearModel class with line-by-line comments

````python
class LinearModel:
    def __init__(my , w , b):
        my.w = w
        my.b = b

    def predict(my , x):
        wx= my.w * x
        
        y= wx + my.b

        return y

model = LinearModel(w=2 , b=1)

result = model.predict(5)
print("Prediction : " , result)
````

# 🎯 Cost Function (Why we need it) — Super Simple

A model can be right or wrong. We need a number that tells us how wrong. That number is called cost (or loss).

**Error for one point:** error = predicted − actual
<br>
If predicted is far from actual → error is big → model is not good.

**Marks example**
Actual = 80, Predicted = 60 → error = -20 (distance is 20).
<br>
We don’t want negative/positive confusion, so we use squared error in MSE.

# MSE (Simple Python) — “Loop style”
<div style="background:blue; color:white; font:bold;"> 
<p style="color:red;">MSE idea in plain words:</p>

1) For every data point, find error
2) Square the error (make it positive and bigger for big mistakes)
3) Add all squared errors
4) Divide by total points (average)

</div>

````python
def mean_squared_error(y_actual , y_pred):
    m = len(y_actual)

    total = 0

    for i in range (m):
        error  = y_pred[i] - y_actual[i]
        sq_error = error ** 2

        total += sq_error

    mse = total / m
    return mse

y_actual = [3, 5, 7]
y_pred   = [2.5, 5.5, 6]
print("MSE:", mean_squared_error(y_actual, y_pred))
````

# 🔗 Connect Model + MSE (Full Pipeline)
```🧪Flow:x → predict → MSE```

**Step-by-step:**
1) Create model (choose w,b)
2) Predict outputs for all x values
3) Compare with actual y values using MSE

````python
class LinearModel:
    def __init__(self, w, b):
        self.w = w
        self.b = b

    def predict(self, x):
        return self.w * x + self.b

def mean_squared_error(y_actual, y_pred):
    n = len(y_actual)
    total = 0
    for i in range(n):
        total += (y_pred[i] - y_actual[i]) ** 2
    return total / n

# Dataset
x = [1, 2, 3, 4, 5]
y_actual = [3, 5, 7, 9, 11]   # true: y = 2x + 1

# Try any model
model = LinearModel(w=1.4, b=1.2)

# Make predictions
y_pred = []
for xi in x:
    y_pred.append(model.predict(xi))

# Calculate cost
cost = mean_squared_error(y_actual, y_pred)

print("Predictions:", y_pred)
print("Cost (MSE):", cost)
````

````python
class LinearModelNumPy:
    def __init__(my , w , b):
        my.w = w
        my.b = b

    def pred(my , x):
        return my.w * x + my.b

    def mse(my  , y , y_pred):
        return np.mean((y_pred - y) ** 2)

x = np.array([1, 2, 3, 4, 5] , dtype=float)
y_actual = np.array([3, 5, 7, 9, 11] , dtype=float)   

model = LinearModelNumPy(w=1.4 , b=1.2)

y_pred = model.pred(x)

cost = model.mse( y_actual, y_pred)

print("Predictions:", y_pred)
print("Cost (MSE):", cost)
````
