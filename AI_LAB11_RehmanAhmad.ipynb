# 📊 Matplotlib Plot Types (Summary Table)

| نمبر | پلاٹ کا نام       | کام (Purpose)                                   | کب استعمال کریں؟                                      | ڈھانچہ / مثال                         |
|------|-------------------|--------------------------------------------------|--------------------------------------------------------|----------------------------------------|
| 1    | Line Plot         | ٹائم سیریز یا رجحان (Trend) دکھاتا ہے          | جب دیکھنا ہو کہ ڈیٹا وقت کے ساتھ کیسے بدل رہا ہے      | `plt.plot(x, y)`                       |
| 2    | Scatter Plot      | دو ویری ایبلز کے درمیان تعلق (Relationship)    | کیا بھی بڑھتا ہے y بڑھنے سے x   | `plt.scatter(x, y)`                    |
| 3    | Bar Plot          | کیٹیگوریکل ڈیٹا کا موازنہ                      | مختلف چیزوں کی مقدار/سکور/فروخت کا موازنہ کرنا ہو     | `plt.bar(categories, values)`          |
| 4    | Histogram         | ڈیٹا کی تقسیم (Distribution) دکھاتا ہے         | معلوم کرنا ہو کہ ڈیٹا زیادہ تر کن ویلیوز کے قریب ہے    | `plt.hist(data, bins=10)`              |
| 5    | Box Plot          | ڈیٹا کی سپریڈ، میڈین، آؤٹ لائرز دکھاتا ہے      | جب ڈیٹا کی variability یا outliers دیکھنے ہوں          | `plt.boxplot(data)`                    |
| 6    | Pie Chart         | کل میں ہر کیٹیگری کا فیصد حصہ دکھاتا ہے        | معلوم کرنا ہو کہ کون سی چیز سب سے بڑی/چھوٹی ہے         | `plt.pie(values, labels=labels)`       |
| 7    | Heatmap           | میٹرکس یا ٹیبل کو رنگوں کی شکل میں دکھاتا ہے   | کوریلیشن، patterns یا گرم/ٹھنڈے علاقے دیکھنے ہوں      | `plt.imshow(matrix, cmap='hot')`       |
| 8    | Subplots          | ایک فگر میں کئی پلاٹس ساتھ دکھاتا ہے           | جب مختلف ڈیٹا سیٹس کا موازنہ کرنا ہو                   | `fig, axs = plt.subplots(2, 2)`        |


 | # | Plot Type         | Purpose                                      | When to Use?                                      | Code Structure                     |
|---|-------------------|----------------------------------------------|---------------------------------------------------|------------------------------------|
| 1 | **Line Plot**     | Show trends over time or sequence            | Tracking changes (e.g., stock price, temperature) | `plt.plot(x, y)`                   |
| 2 | **Scatter Plot**  | Show relationship between two variables       | Finding correlations or patterns                  | `plt.scatter(x, y)`                |
| 3 | **Bar Plot**      | Compare categorical data                     | Comparing values across categories (e.g., cities) | `plt.bar(categories, values)`      |
| 4 | **Histogram**     | Show distribution of numerical data          | Understanding data spread (e.g., exam scores)     | `plt.hist(data, bins=10)`          |
| 5 | **Box Plot**      | Show median, spread, and outliers            | Analyzing data variability & anomalies            | `plt.boxplot(data)`                |
| 6 | **Pie Chart**     | Show parts of a whole (percentages)          | Displaying proportions (e.g., budget allocation)  | `plt.pie(values, labels=labels)`   |
| 7 | **Heatmap**       | Visualize 2D data using colors               | Showing matrices (e.g., correlation, sensors)     | `plt.imshow(matrix, cmap='hot')`   |
| 8 | **Subplots**      | Display multiple plots in one figure         | Comparing different datasets side by side         | `fig, axs = plt.subplots(2, 2)`    |

```python
hours = np.array([1,2,3,4,5,6])
marks = np.array([50,55,60,63,70,78])

m , c = np.polyfit(hours , marks ,1)
print(f"m= {m} , c = {c}")

# np.polyfit(x, y, 1) → ایک سیدھی لکیر (best-fit line) نکالتا ہے جو ڈیٹا کے قریب ترین ہو۔
# 1 کا مطلب: صرف linear (degree 1) equation — یعنی y = mx + c


predicted = m*hours + c
print(predicted)
plt.scatter(hours , marks)
plt.plot(hours , predicted , color="red")
plt.title("Linear Regression - Marks Prediction")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

```python
hours = np.array([1,2,3,4,5,6,7])
marks = np.array([50,55,60,63,70,78,85])

m , c = np.polyfit(hours , marks ,1)

predicted = m*hours + c

plt.scatter(hours , marks)
plt.plot(hours , predicted , color="red")
plt.title("Linear Regression - Marks Prediction")
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

```python
marks = np.array([35,50,72,28,65,90])
labels = []

for m in marks:
    if m >=40:
        labels.append("Pass")
    else:
        labels.append("Fail")

print("Marks : " , marks)
print("Labels : " , labels)
```

```python
marks = np.array([35,50,72,28,65,90,34,56,78,43,12])
labels = []

for m in marks:
    if m >=40:
        labels.append("Pass")
    else:
        labels.append("Fail")

print("Marks : " , marks)
print("Labels : " , labels)


index = np.arange(0,len(marks) , 1)
print(index)
# plt.plot(index , marks , marker = "o")
plt.bar(index ,marks ,label='Pass/Fail Threshold')
plt.title("Marks labels(Pass , Fail)")
plt.xlabel("Index")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)
plt.show()



plt.plot(marks, 'o-', color='blue')  # 'o' = dots, '-' = line
plt.axhline(y=40, color='red', linestyle='--', label='Pass/Fail Threshold')
plt.xlabel('Student Index')
plt.ylabel('Marks')
plt.title('Student Marks with Pass/Fail Threshold')
plt.legend()
plt.grid(True)
plt.show()
```
